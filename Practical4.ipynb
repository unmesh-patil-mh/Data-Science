```python
from sklearn.model_selection import train_test_split
```

```python
import sklearn
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lm = LinearRegression()
```

```python
model=lm.fit(xtrain, ytrain)
```

```python
ytrain_pred = lm.predict(xtrain)
```

```python
ytest_pred = lm.predict(xtest)
```
